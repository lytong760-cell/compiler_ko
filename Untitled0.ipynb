```python
from google.colab import userdata
import subprocess

# 1. Lấy secret URL từ Colab
repo_url = userdata.get('REPO_URL')

# 2. Viết script bash với biến đã được điền sẵn giá trị
# Dùng f-string để thay thế $REPO_URL bằng giá trị thật ngay trong Python
bash_script = f"""
#!/bin/bash

npm install -g @kilocode/cli
sudo apt update && sudo apt install -y htop


git config --global user.email "lytong660-cell@users.noreply.github.com"
git config --global user.name "lytong660-cell"

echo "🔄 Đang clone repository..."
# Biến này giờ đã có giá trị thật, không còn là $REPO_URL rỗng nữa
git clone "{repo_url}" || {{ echo "❌ Clone thất bại!"; exit 1; }}

cd compiler_ko || {{ echo "❌ Không tìm thấy thư mục 'compiler_ko'."; exit 1; }}

echo "✅ Đã vào thư mục làm việc. Bắt đầu auto-commit..."

while true; do
    if [ -n "$(git status --porcelain)" ]; then
        git add .
        COMMIT_MSG="Auto save: $(date +'%Y-%m-%d %H:%M:%S')"
        git commit -m "$COMMIT_MSG"
        git push origin main || echo "[WARN] Push thất bại lúc $(date)"
    fi
    sleep 15
done
"""

# 3. Chạy script
print("🚀 Đang khởi chạy script auto-commit...")
process = subprocess.Popen(bash_script, shell=True, executable='/bin/bash')
```
