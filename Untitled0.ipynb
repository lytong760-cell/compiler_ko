```python
%%bash
#!/bin/bash

# ==========================================
# CẤU HÌNH GIT IDENTITY
# ==========================================
git config --global user.email "lytong660-cell@users.noreply.github.com"
git config --global user.name "lytong660-cell"

# ==========================================
# CÀI ĐẶT CLI TOOL (Nếu chưa có)
# ==========================================
# npm install -g @kilocode/cli # Bỏ comment nếu cần cài lại

# ==========================================
# ⚠️ CẤU HÌNH REPO URL (SỬA CH NÀY)
# ==========================================
# KHUYẾN NGHỊ: Dùng SSH để tránh hết hạn token
# Lấy URL SSH tại: GitHub Repo -> Code -> SSH


# Nếu bắt buộc dùng HTTPS, hãy đảm bảo token còn hạn và đúng quyền 'repo'
# REPO_URL="https://ghp_XXXXXXXXXXXXXXXXXXXX@github.com/lytong660-cell/compiler_ko_git.git"

# ==========================================
# CLONE VÀ KIỂM TRA THƯ MỤC
# ==========================================
echo "🔄 Đang clone repository..."
git clone "$REPO_URL" || { echo "❌ Clone thất bại! Kiểm tra lại URL hoặc SSH Key."; exit 1; }

cd compiler_ko || { echo "❌ Không tìm thấy thư mục 'compiler_ko'. Kiểm tra tên repo."; exit 1; }

echo "✅ Đã vào thư mục làm việc. Bắt đầu auto-commit..."

# ==========================================
# VÒNG LẶP AUTO-COMMIT & PUSH
# ==========================================
while true; do
    # Kiểm tra xem có file nào thay đổi không
    if [ -n "$(git status --porcelain)" ]; then
        git add .

        # Commit với timestamp chính xác
        COMMIT_MSG="Auto save: $(date +'%Y-%m-%d %H:%M:%S')"
        git commit -m "$COMMIT_MSG"

        # Push lên main, nếu lỗi sẽ cảnh báo nhưng không dừng script
        git push origin main || echo "[️ WARN] Push thất bại lúc $(date)"
    fi

    # Chờ 15 giây trước khi kiểm tra lần tiếp theo
    sleep 15
done
```
